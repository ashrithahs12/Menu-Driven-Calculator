```python
Project title:Console-Based Smart Calculator
Problem statement:
    The project aims to create a menu-driven smart calculator usingpython and NumPy that allows users to perform different
mathematical operations,display results,and continue or exit through a console interface.
Logic explanation:
    This program uses a menu-driven approach to perform arithmetic operations and NumPy array addition based on the users
choice.It repeatedly accepts input, calls the corresponding function to perform the operation, and stops only when the user selects the Exit option.
```

```python
import numpy as np
def addition(a,b):
    return a+b
def subtraction(a,b):
    return a-b
def multiplication(a,b):
    return a*b
def division(a,b):
    if b==0:
        print("Division by zero is not possible")
    else:
        return a/b
def floor_division(a,b):
    return a//b
def power(a,b):
    return a**b
def modulus(a,b):
    if b==0:
        print("Division by zero is not possible")
    else:
        return a%b
def numpy_array_addition():
  n=int(input("Enter number of elements: "))
  list1=[]
  list2=[]
  print("Enter elements for Array1")
  for i in range(n):
    value=int(input())
    list1.append(value)
  print("Enter elements for Array2")
  for i in range(n):
    value=int(input())
    list2.append(value)
  array1=np.array(list1)
  array2=np.array(list2)
  result=array1+array2
  print("Array1=",array1)
  print("Array2=",array2)
  print("Result=",result)
choice=0
while choice!=9:
  print("\n----------Menu driven calculator--------------")
  print("1.Addition")
  print("2.Subtraction")
  print("3.Multiplication")
  print("4.Division")
  print("5.Floor Division")
  print("6.Power")
  print("7.Modulus")
  print("8.Numpy array addition")
  print("9.Exit")
  choice=int(input("Enter your choice: "))
  if choice>=1 and choice<=7:
      num1=int(input("Enter first number: "))
      num2=int(input("Enter second number: "))
  if choice==1:
     print("Addition=",addition(num1,num2))
  elif choice==2:
     print("Subtraction=",subtraction(num1,num2))
  elif choice==3:
     print("Multiplication=",multiplication(num1,num2))
  elif choice==4:
     print("Division=",division(num1,num2))
  elif choice==5:
     print("Floor Division=",floor_division(num1,num2))
  elif choice==6:
     print("Power=",power(num1,num2))
  elif choice==7:
     print("Modulus=",modulus(num1,num2))
  elif choice==8:
    numpy_array_addition()
  elif choice==9:
    print("Program Ended")
  else:
    print("Invalid choice")

```
